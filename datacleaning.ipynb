```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "name": ["Amit", "Rahul", "Priya", "Amit", "Neha"],
    "age": [25, None, 28, 25, 35],
    "salary": [50000, 70000, None, 50000, 90000],
    "city": ["Mumbai", "Pune", "Mumbai", "Mumbai", None]
})
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
df.duplicated()
```

```python
df.drop_duplicates()
```

```python
df["city"] = df["city"].fillna("unknown")
```

```python
df["salary"] = df["salary"].fillna(df["salary"].median())
```
