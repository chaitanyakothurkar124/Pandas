```python
import pandas as pd
```

```python
import numpy as np
```

```python


df = pd.DataFrame({
    "name": ["Amit", "Rahul", "Priya", "Neha", "Vikram"],
    "age": [25, 32, 28, 35, 24],
    "salary": [50000, 75000, 65000, 90000, 45000],
    "department": ["IT", "Finance", "IT", "HR", "Finance"]
})

df
```

```python
df["bonus"] = df["salary"] * 0.15
```

```python
df["total_salary"] = df["salary"] + df["bonus"]
```

```python
df
```

```python
df["salary_level"] = df["salary"].apply(
    lambda x: "High" if x >= 70000 else "Normal"
)
```

```python
df
```

```python
department_codes = {
    "IT": "I",
    "Finance": "F",
    "HR": "H"
}
df["department_codes"] = df["department"].map(department_codes)
```

```python
df
```

```python
df["age_group"] = df["age"].apply(
    lambda age: "Young" if age < 30 else "30+"
)
```

```python
df
```

```python
df
```

```python
conditions = [
    df["salary"] < 50000,
    df["salary"].between(50000, 69999),
    df["salary"] >= 70000
]

choices = [
    "Low",
    "Medium",
    "High"
]
```

```python
df["salary_level"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```
